```python
# getting a subset
from fairchem.core.datasets import LmdbDataset
import random
from ase.io import read, write
from ase import Atoms
```

```python
def data_to_atoms(data):
    numbers = data.atomic_numbers
    positions = data.pos
    cell = data.cell.squeeze()
    atoms = Atoms(numbers=numbers.cpu().detach().numpy(), 
                  positions=positions.cpu().detach().numpy(), 
                  cell=cell.cpu().detach().numpy(),
                  pbc=[True, True, True])
    return atoms
```

```python
cfg = {'src': '/data/shared/MLFF/OC20/s2ef/2M/train'}
dataset = LmdbDataset(cfg)
len(dataset)
```

```python
# choose 50 random atoms from this dataset
random.seed(88)
random_indices = random.sample(range(len(dataset)), 100)
random_indices
```

```python
# now need to get this subset and do the descriptors
structures = []
for i in random_indices:
    structures.append(data_to_atoms(dataset[i]))
print(len(structures))
```

```python
structures[0]
```

```python
write('oc20/oc20_subset.extxyz', structures)
```

```python
structures_read = read('oc20/oc20_subset.extxyz', index=":")
len(structures_read)
```

```python
structures_read[0]
```

```python
# now need to get the rest of the dataset without this subset
all_other_structures = []
for i in range(len(dataset)):
    if i not in random_indices:
        all_other_structures.append(data_to_atoms(dataset[i]))
print(len(all_other_structures))
```

```python
write('oc20/oc20_without_subset.extxyz', all_other_structures)
```
